```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Cargamos los datos

```python
# Carga el archivo CSV en un DataFrame de pandas
# 'parse_dates' convierte automáticamente la columna 'Month' a formato fecha (datetime)
df = pd.read_csv('data/df_furniture.csv', parse_dates=['Month'])

# Muestra las primeras 5 filas del DataFrame para verificar que los datos cargaron bien
df.head()
```

## 2. Tratar la fecha como texto e interpretar (parse)

```python
df.rename(columns={'Month' : 'date'}, inplace=True)
```

```python
# Rellena los valores faltantes (NaN) en la columna 'date' usando interpolación
# El método 'linear' asume un espacio constante entre los puntos para calcular los valores huecos
df['date'] = df['date'].interpolate(method='linear')
```

```python
# Genera un rango de fechas completo y lo asigna a la columna 'date'
# Comienza en la fecha mínima y termina en la máxima, asegurando que sea el primer día de cada mes ('MS')
df['date'] = pd.date_range(df['date'].min(), df['date'].max(), freq='MS')
```

```python
# Crea una nueva columna llamada 'fecha_str' convirtiendo los valores de 'date' a texto (string)
# El método .astype(str) transforma el objeto tipo datetime en una cadena de caracteres
df['fecha_str'] = df['date'].astype(str)
```

```python
# Divide la columna de texto 'fecha_str' en tres nuevas columnas usando el guion '-' como separador
# 'expand=True' convierte el resultado directamente en un DataFrame de 3 columnas
df[['year', 'month', 'day']] = df['fecha_str'].str.split('-', expand=True)

# Convierte las nuevas columnas de texto (string) a números enteros (int)
# Esto es necesario para poder realizar cálculos matemáticos o filtrar por rangos numéricos
df['year'] = df['year'].astype(int)
df['month'] = df['month'].astype(int)
df['day'] = df['day'].astype(int)
```

```python
# Reconstruye una columna de fecha completa a partir de las columnas individuales de año, mes y día
# pd.to_datetime busca automáticamente nombres de columnas como 'year', 'month' y 'day' para ensamblarlas
df['new_date'] = pd.to_datetime(df[['year', 'month', 'day']])

# Muestra las primeras 5 filas para verificar que la nueva columna 'new_date' se creó correctamente
df.head()
```

## 3. ¿De qué fecha a qué fecha va el histórico?

```python
# Se asegura de que la columna 'date' tenga el formato de fecha (datetime) oficial de pandas
# Esto es útil si alguna operación anterior la hubiera convertido accidentalmente a otro tipo
df['date'] = pd.to_datetime(df['date'])
```

```python
# Imprime un mensaje en la consola mostrando el rango temporal del conjunto de datos
# df['date'].min() obtiene la fecha más antigua y .max() la más reciente
print('La fecha mínima es ', df['date'].min(), 'y la máxima ', df['date'].max())
```

## 4. ¿Cómo es la serie temporal? ¿Tiene tendencia? ¿Aprecias alguna estacionalidad?

```python
# Genera un gráfico de línea usando la columna 'date' para el eje X y 'furniture_price_adjusted' para el eje Y
# El argumento '-r' define el estilo de la línea: una línea sólida ('-') de color rojo ('r')
# El punto y coma (;) al final se usa en entornos como Jupyter para evitar que se imprima el texto del objeto del gráfico
plt.plot(df['date'], df['furniture_price_adjusted'], '-r');
```

## 5. ¿Podrías demostrar de dos maneras estadísticas que la serie no es estacionaria?

```python
# Importa las pruebas estadísticas para series temporales: Augmented Dickey-Fuller (ADF) y KPSS
from statsmodels.tsa.stattools import adfuller, kpss

# Ejecuta la prueba de Dickey-Fuller aumentada sobre los precios de los muebles
# Se usa .values para pasar solo los números sin los índices de pandas
result = adfuller(df['furniture_price_adjusted'].values)

# Accede al segundo elemento del resultado (índice 1), que es el p-valor (p-value)
result[1]
```

```python
# Calcula y muestra la media y la mediana de los primeros 20 registros (inicio de la serie)
print(df['furniture_price_adjusted'].iloc[:20].agg(['mean','median']))

# Calcula y muestra la media y la mediana de un bloque intermedio (filas 80 a 149)
print(df['furniture_price_adjusted'].iloc[80:150].agg(['mean','median']))

# Calcula y muestra la media y la mediana de un bloque final (filas 230 a 299)
print(df['furniture_price_adjusted'].iloc[230:300].agg(['mean','median']))
```

## 6. ¿Conoces algún método para representar mejor la tendencia?

```python
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# Realiza la descomposición de la serie temporal usando un modelo multiplicativo
# 'period=12' indica que los datos tienen un ciclo de 12 meses (estacionalidad anual)
result_multi = seasonal_decompose(df['furniture_price_adjusted'], model='multiplicative', period=12)

# Desestacionaliza los datos restando el componente estacional (la variación cíclica anual)
# Se usa .values para obtener los números puros y restarlos del precio ajustado
deseasonalized = df.furniture_price_adjusted.values - result_multi.seasonal

# Grafica la serie resultante sin el efecto de las estaciones
plt.plot(deseasonalized)

# Añade un título al gráfico especificando que el precio está desestacionalizado
plt.title('Furniture price deseasonalized', fontsize=16);
```

```python
# Grafica únicamente el componente de la tendencia extraído por la descomposición
# La tendencia muestra el movimiento a largo plazo de la serie, eliminando el ruido y la estacionalidad
plt.plot(result_multi.trend);
```

## 7. Estaría bien visualizar todos los componentes de la serie temporal por separado

```python
# Genera automáticamente cuatro gráficos en uno solo:
# 1. Observed: Los datos originales.
# 2. Trend: La dirección a largo plazo (tendencia).
# 3. Seasonal: El patrón que se repite cada 12 meses (estacionalidad).
# 4. Resid: El "ruido" o lo que queda que no se explica por tendencia ni estacionalidad.
result_multi.plot()
```

```python
# Grafica el componente residual (el ruido) de la descomposición
# Representa lo que queda tras eliminar la tendencia y la estacionalidad de los datos originales
plt.plot(result_multi.resid);
```

## 8. Demuestra mediante una gráfica de autocorrelación que la serie es estacional

```python
from pandas.plotting import autocorrelation_plot

# Genera un gráfico de autocorrelación (correlograma) para los precios de los muebles
# Ayuda a identificar patrones repetitivos y ver si los valores pasados influyen en los futuros
autocorrelation_plot(df.furniture_price_adjusted.values);
```

```python
from statsmodels.graphics.tsaplots import plot_acf
# Genera un gráfico de la Función de Autocorrelación (ACF)
# Muestra la correlación entre la serie y sus propios valores pasados (lags)
plot_acf(df['furniture_price_adjusted']);
```

## 9. ¿Qué dependencia tiene cada instante de sus últimos 5 lags? Demuéstralo mediante una matriz de correlación.

```python
# Genera el gráfico de la Función de Autocorrelación (ACF) limitado a los primeros 25 retrasos (lags)
# Esto permite observar con más detalle el comportamiento de los dos primeros años de datos
plot_acf(df['furniture_price_adjusted'], lags=25);
```

```python
# Creamos columnas para los últimos 5 lags
for lag in range(1, 13):
    df[f'lag_{lag}'] = df['furniture_price_adjusted'].shift(lag)

# Calculamos la matriz de correlación
correlation_matrix = df[['furniture_price_adjusted'] + [f'lag_{lag}' for lag in range(1, 13)]].corr()

# Visualizamos la matriz de correlación
plt.figure(figsize=(20,10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Matriz de Correlación con Lags')
plt.show()
```

## 11. Para un futuro modelado de los datos necesitaremos eliminar la tendencia. ¿Cómo hacemos esto?

```python
from statsmodels.tsa.seasonal import seasonal_decompose

# Aplica la descomposición multiplicativa (asume que los ciclos crecen con la tendencia)
result_multi = seasonal_decompose(df['furniture_price_adjusted'], model='multiplicative', period=12)

# Elimina la tendencia dividiendo el valor original por el componente de tendencia
# Al ser un modelo multiplicativo, se usa división (en lugar de resta) para "limpiar" la serie
detrended = df.furniture_price_adjusted.values / result_multi.trend

# Grafica el resultado para observar la serie "plana" (solo estacionalidad y ruido)
plt.plot(detrended)

# Añade un título descriptivo con un tamaño de fuente de 16
plt.title('Precio de los muebles ajustado sin tendencia', fontsize=16);
```

```python
# Grafica la serie tras aplicar una diferenciación (restar a cada valor el anterior)
# Esto sirve para eliminar la tendencia y estabilizar la media
plt.plot(df.furniture_price_adjusted.diff())

# Realiza la prueba de Dickey-Fuller (ADF) sobre la serie diferenciada
# .dropna() es vital porque la primera fila se convierte en NaN al no tener un valor previo que restar
print(adfuller(df.furniture_price_adjusted.diff().dropna().values)[1])

# Añade el título indicando que la serie se ha desestacionalizado/detrended mediante diferenciación
plt.title('Drug Sales detrended by differencing values', fontsize=16);
```
